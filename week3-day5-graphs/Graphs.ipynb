# Graphs:

A graph is a **non-linear data structure** made up of **vertices** (nodes) and **edges** (connections) that represent relationships between objects. Unlike arrays or linked lists, graphs do not follow a sequential order.



# Types:

**Based on Weight**
1. Weighted Graphs
2. Unweighted Graphs

**Based on Edge Direction**

3. Undirected Graph / Bi-Directional
4. Directed Graph / Uni-Directional

# Representation of Graph Data Structure:
1. Adjacency List Representation of Graph
2. Adjacency Matrix Representation of Graph Data Structure

# Real-Life Applications of Graph Data Structure:

*   Social Networks
*   Computer Networks / Internet
*   Neural Networks

and many more...




# Advantages of Graph Data Structure:


*   Graphs are flexible
*   Model real-world problems
*   Represent items and relationships
*   Simplifies complex data





```python
from collections import deque

class Graph:
    def __init__(self):
        self.graph = {}

    def add_vertex(self, vertex):
        if vertex not in self.graph:
            self.graph[vertex] = []

    def add_edge(self, u, v):
        # undirected graph
        self.graph[u].append(v)
        self.graph[v].append(u)

    def display(self):
        for vertex in self.graph:
            print(vertex, "->", self.graph[vertex])

    def remove_edge(self, u, v):
        if v in self.graph[u]:
            self.graph[u].remove(v)
        if u in self.graph[v]:
            self.graph[v].remove(u)

    def remove_vertex(self, vertex):
        for neighbor in self.graph[vertex]:
            self.graph[neighbor].remove(vertex)
        del self.graph[vertex]

    #  Depth First Search
    def dfs(self, start, visited=None):
        if visited is None:
            visited = set()

        visited.add(start)
        print(start, end=" ")

        for neighbor in self.graph[start]:
            if neighbor not in visited:
                self.dfs(neighbor, visited)

    # Breadth First Search
    def bfs(self, start):
        visited = set()
        queue = deque([start])
        visited.add(start)

        while queue:
            node = queue.popleft()
            print(node, end=" ")

            for neighbor in self.graph[node]:
                if neighbor not in visited:
                    visited.add(neighbor)
                    queue.append(neighbor)
```

```python
g = Graph()

g.add_vertex("A")
g.add_vertex("B")
g.add_vertex("C")
g.add_vertex("D")

g.add_edge("A", "B")
g.add_edge("A", "C")
g.add_edge("B", "D")
g.add_edge("C", "D")

g.display()
```

```python
g.dfs("A")
```

```python
g.bfs("A")
```
