# Methods using Adjacency Matrix


Methods:
1. Add vertex
2. Add edge
3. Remove vertex
4. Remove edge

```python
class Graph:
  def __init__(self):
    self.vertices = [] # vertices
    self.matrix = [] # matrix for edges

  def add_vertex(self, vertex):
    if vertex not in self.vertices:
      self.vertices.append(vertex) # vertex added

      # mark edges with other vertices
      for row in self.matrix:
        row.append(0) # currently no edge

      # rows of 0 for this vertex
      zeroes = [0] * len(self.vertices)
      self.matrix.append(zeroes)

  def add_edge(self, vertex1, vertex2):
    # get index of vertices
    i = self.vertices.index(vertex1)
    j = self.vertices.index(vertex2)

    # connect them
    self.matrix[i][j] = 1
    self.matrix[j][i] = 1

  def display(self):
    print("vertices: ", self.vertices)
    print("edges connection")
    n = len(self.vertices)
    for i in range(n):
      print(i, self.matrix[i])
    print("\n")

  def remove_edge(self, vertex1, vertex2):
    # get index of vertices
    i = self.vertices.index(vertex1)
    j = self.vertices.index(vertex2)

    # disconnect them
    self.matrix[i][j] = 0
    self.matrix[j][i] = 0

  def remove_vertex(self, vertex):
    index = self.vertices.index(vertex)

    self.vertices.pop(index) # delete vertex
    self.matrix.pop(index) # delete row
    # delete column
    for row in self.matrix:
      row.pop(index)
```

```python
friends = Graph()

friends.add_vertex("faraz")
friends.add_vertex("abdullah")
friends.add_vertex("ebad")
friends.add_vertex("zubair")

print(friends.matrix)

friends.display()

friends.add_edge("faraz", "abdullah")
friends.display()

friends.add_edge("ebad", "zubair")
friends.display()

friends.add_edge("ebad", "abdullah")
friends.display()


```

# Time and Space complexity

1. Add vertex
  * Time: O(V)
  * Space: O(V)
2. Add edge
  * Time: O(V)
  * Space: O(1)
3. Remove vertex
  * Time: O(V)
  * Space: O(1)
4. Remove edge
  * Time: O(V)
  * Space: O(1)

Space Complexity of Graph: O(V^2)
