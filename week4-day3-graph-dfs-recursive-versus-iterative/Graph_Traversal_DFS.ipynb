Depth Fist Search (DFS) - Recursive versus Iterative

```python
# Copy past this code on Python Tutor
# https://pythontutor.com/visualize.html#mode=edit

class Graph:
    def __init__(self):
        # adjacency list representation
        self.graph = {}

    def add_vertex(self, vertex):
        if vertex not in self.graph:
            self.graph[vertex] = []

    def add_edge(self, u, v):
        # undirected graph
        if u not in self.graph:
            self.graph[u] = []
        if v not in self.graph:
            self.graph[v] = []

        self.graph[u].append(v)
        self.graph[v].append(u)

    # ----------------------------
    # Recursive DFS
    # ----------------------------
    def dfs_recursive(self, start, visited=None):
        if visited is None:
            visited = set()

        visited.add(start)
        print(start, end=" ")

        for neighbor in self.graph[start]:
            if neighbor not in visited:
                self.dfs_recursive(neighbor, visited)

    # ----------------------------
    # Iterative DFS
    # ----------------------------
    def dfs_iterative(self, start):
        visited = set()
        stack = []

        stack.append(start)

        while stack:
            node = stack.pop()

            if node not in visited:
                visited.add(node)
                print(node, end=" ")

                # push neighbors in reverse order
                # so traversal order matches recursive DFS
                for neighbor in reversed(self.graph[node]):
                    if neighbor not in visited:
                        stack.append(neighbor)


# ----------------------------
# Example usage (Python Tutor friendly)
# ----------------------------
g = Graph()

g.add_edge(0, 1)
g.add_edge(0, 2)
g.add_edge(1, 3)
g.add_edge(1, 4)
g.add_edge(2, 5)

print("Recursive DFS:")
g.dfs_recursive(0)

print("\n\nIterative DFS:")
g.dfs_iterative(0)
```
