List (Array) Implementation

Methods:
* append(): add an element to the end
* pop(): remove and return the last element
* get(index): return the value stored at index
* size(): return the size of list

This is a dynamic array, it will resize when its full and an elements needs to be added

```python
class MyList:
  def __init__(self, capacity=4):
    self.capacity = capacity  # total allocated space
    self.n = 0  # current number of elements
    self.arr = [None] * capacity  # fixed-size array
# [0], [n=1] [x,1]

  def _resize(self):
    new_capacity = self.capacity * 2 # 8
    new_arr = [None] * new_capacity

    for i in range(self.n): # 4 times, 0-3
      new_arr[i] = self.arr[i]

    # del self.arr
    self.arr = new_arr
    self.capacity = new_capacity



  def append(self, x):
    if self.n == self.capacity:
      self._resize()

    self.arr[self.n] = x # [1, x,N,N]
    self.n += 1 # n = 2

  # [1, 2, N, N] -> n=2
  # 1, n-1
  def pop(self):
    if self.n == 0:
      raise IndexError("pop from empty list")

    value = self.arr[self.n - 1]
    self.arr[self.n - 1] = None
    self.n -= 1

    return value

  # [1, N, N, N] n=1
  # 1,2,3
  def get(self, index):
    if index < 0 or index >= self.n:
      raise IndexError("index out of bounds")

    return self.arr[index]

  # [1, N, N, N] n=1
  # 1 return value
  def size(self):
    return self.n


  def display(self):
    # print(self.arr)
    print(self.arr, "\n")
```

```python
# create list
my_list = MyList()
print("initial list")
my_list.display()
# print(my_list)



```

```python
# create list
my_list = MyList()
print("initial list")
my_list.display()

# append elements
my_list.append(10)
my_list.append(20)
my_list.append(30)
print("elements added")
my_list.display()

# get element
value = my_list.get(1)
print("element at index 1:", value)

# list size
size = my_list.size()
print("size:", size)

# pop element
popped = my_list.pop()
print("popped element:", popped)
my_list.display()

# append more elements to trigger resize
my_list.append(40)
my_list.append(50)
print("full list")
my_list.display()

my_list.append(60)  # This will trigger _resize internally
print("resized list")
my_list.display()

```
