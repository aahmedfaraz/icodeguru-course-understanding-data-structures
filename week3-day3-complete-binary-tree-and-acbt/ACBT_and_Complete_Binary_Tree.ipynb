**Binary Tree Variations:**

1.   Binary Tree
2.   Full or Strict Binary Tree
3.   Almost Complete Binary Tree (ACBT)
4.   Complete Binary Tree (***VERY IMPORTANT***)
5.   Binary Search Tree



**Complete Binary Tree:**
A Complete Binary Tree is a binary tree where:

1. All levels are completely filled,
2. Except possibly the last level
3. The last level is filled from left to right

**Key Properties**

- Height is minimum possible
- Very compact
- Easy to store in array

**Index Rules** when Complete Binary Tree is stored in form of array:
- Parent: (i - 1) // 2
- Left child: 2i + 1
- Right child: 2i + 2

Benefits of CBT or ACBT:
1. **Minimum Height** → Faster
    *   Height ≈ log₂(n)
    *   Operations = O(log n)
2. **Perfect Fit for Arrays** (No Pointers Needed)
    * You can use formulas
    * No extra space needed to store pointers
3. **No Wasted Space** - like Sparse Tree Problem
4. Predictable Insert & Delete
    * Insert → next available leftmost position
    * Delete → replace with last node


```python
class BinaryTreeArray:
    """
    Base class to represent a binary tree using array representation.
    """
    def __init__(self):
        self.tree = []

    # Insert node (left to right)
    def insert(self, value):
        self.tree.append(value)

    # Get left child index
    def left(self, i):
        return 2 * i + 1

    # Get right child index
    def right(self, i):
        return 2 * i + 2

    # Display tree as array
    def display(self):
        print(self.tree)


# ------------------------------------------------
# Complete Binary Tree
# (Also called Almost Complete Binary Tree in some books)
# ------------------------------------------------
class CompleteBinaryTree(BinaryTreeArray):
    """
    Complete Binary Tree:
    - All levels are fully filled except possibly the last
    - Last level is filled from left to right

    NOTE:
    Some books call this an 'Almost Complete Binary Tree'.
    They are the SAME thing.
    """

    def is_complete(self):
        """
        Once we see an empty spot (None),
        there should be no more real nodes after it.
        """
        found_none = False

        for value in self.tree:
            if value is None:
                found_none = True
            elif found_none:
                return False

        return True


# -----------------------------
# DEMO
# -----------------------------
if __name__ == "__main__":

    print("=== Complete Binary Tree Example ===")
    cbt1 = CompleteBinaryTree()

    for v in [1, 2, 3, 4, 5]:
        cbt1.insert(v)

    cbt1.display()
    print("Is Complete Binary Tree?", cbt1.is_complete())


    cbt2 = CompleteBinaryTree()
    for v in [1, 2, 3, 4, None, 5]:
        cbt2.insert(v)

    cbt2.display()
    print("Is Complete Binary Tree?", cbt2.is_complete())
```
